```python
# 3. Removendo linhas onde todas as colunas especificadas são nulas
# (Mantendo a lógica de JEyHNQNbMS7v)
# 'id' não deve ser usado aqui se for a chave primária e não tiver nulos no original
df_branco = df_sem_duplicatas.dropna(subset=colunas_para_duplicatas, how='all')

print(f"Linhas após remover linhas com todas as colunas nulas: {len(df_branco)}")
```

```python
# 4. Removendo valores impossíveis (Mantendo a lógica de 2bfb2447)
df_limpo = df_branco[
    (df_branco['tempo_ciclo'] > 0) &
    (df_branco['refugo_pct'] >= 0) & (df_branco['refugo_pct'] <= 100) &
    (df_branco['temperatura'] >= 0) & (df_branco['temperatura'] <= 200)
]

print(f"Linhas finais em df_limpo após remover valores impossíveis: {len(df_limpo)}")
print("\nVerificando os valores únicos de 'turno' no df_limpo atualizado:")
print(df_limpo['turno'].unique())
```

```python
media_tempo_ciclo_por_turno = df_limpo.groupby('turno')['tempo_ciclo'].mean()
mediana_tempo_ciclo_por_turno = df_limpo.groupby('turno')['tempo_ciclo'].median()

print("Média de tempo_ciclo por turno (df_limpo atualizado):")
print(media_tempo_ciclo_por_turno)
print("\nMediana de tempo_ciclo por turno (df_limpo atualizado):")
print(mediana_tempo_ciclo_por_turno)
```

## quartil

```python
q1 = df['tempo_ciclo'].quantile(0.25)

q3 = df['tempo_ciclo'].quantile(0.75)

iqr = q3 - q1

print('0.25%: ' , round(q1 , 2))
print('0.75%: ' , round(q3 , 2))
print('0.50%: ' , round(iqr, 2))
```

## outliers

```python
#limites

limitebaixo = q1 - 1.5 * iqr
limitealto = q3 - 1.5 * iqr

print('limite inferior ', round(limitebaixo, 2))
print('limite superior' , round(limitealto, 2))
print()
print(f'valores abaixo de {limitebaixo:.2f} ou acima de {limitealto:.2f} são outliers')
```

```python
#encontrado outliers

outliers = df[
    (df['tempo_ciclo'] < limitebaixo) |
    (df['tempo_ciclo'] > limitealto)
]

print('outliers encontrados: ' , len(outliers))
print()
outliers[['id' , 'turno' , 'maquina' , 'tempo_ciclo' , 'operador']]
```

## probabilidade
- quantidade de vezes que algo acontece dividido pelo total de vezes

```python
#quantas peças tem refugo acima de 5%

total = len(df)
acima5 = len(df[df['refugo_pct']>5])

prob = acima5 / total

print(f'peças com refugo >5%: {acima5} de {total}')
print(f'probabilidade: {prob:.0%}')
```
