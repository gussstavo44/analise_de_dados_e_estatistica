```python
#1
import pandas as pd

# carregar o  CSV
df = pd.read_csv('moveis_producao.csv')

#2

# remove linhas com dados vazios
df = df.dropna(how='all', subset=['dia', 'setor', 'produto', 'quantidade', 'defeitos'])

#3

# remover espaços em branco e padronizar a estrutura
df['setor'] = df['setor'].str.strip().str.title()

#4

# remove duplicatas (exceto ID) para contagem
df = df.drop_duplicates(subset=['dia', 'setor', 'produto', 'quantidade', 'defeitos'])

#5

# agrupa por setor e soma a quantidade de defeitos ; depois ordenar do maior para o menor
defeitos_por_setor = df.groupby('setor')['defeitos'].sum().reset_index()

defeitos_por_setor = defeitos_por_setor.sort_values(by='defeitos', ascending=False)
print(defeitos_por_setor)

#6

## remove duplicatas
df = df.drop_duplicates(subset=['dia', 'setor', 'produto', 'quantidade', 'defeitos'])

# mostra o número de linhas restantes
print(f"Linhas restantes: {len(df)}")

#7

# produção por setor
producao_por_setor = df.groupby('setor')['quantidade'].sum().reset_index()
print(producao_por_setor)

#8

# defeitos por setor
defeitos_por_setor = df.groupby('setor')['defeitos'].sum().reset_index()
print(defeitos_por_setor)

#10

# salva em csv
df.to_csv('moveis_producao_limpo.csv', index=False)
```

9: O setor com mais defeitos é o de Pintura, com 29 defeitos no total, tendo uma taxa de defeitos em 12%.
