# remover linha vazia

```python
colunas = [
    "data",
    "turno",
    'codigo_peca',
    'qtd_produzida',
    'qtd_refugo',
    'temperatura_c',
    'status'
]

df = df.dropna(subset=colunas, how='all')
print('linhas agora:', df.shape[0])
```

# padronizacao turnos


```python
from enum import unique
#padronização do turno

df['turno'].unique()

# tira espaço e deixa minusculo

df['turno'] = df['turno'].str.strip().str.lower()

#troca pelos nomes corretos
df['turno'] = df['turno'].replace({
    'manha': 'Manha',
    "manhã": 'Manha',
    'tarde': "Tarde",
    'noite': 'Noite'
})
#conferindo
df['turno'].unique()
```

# padronizacao peças

```python
from enum import unique
#padronizar

df['codigo_peca'] = df['codigo_peca'].str.strip().str.upper()
df['codigo_peca'].unique

#remover duplicados

antes = df.shape[0]
df = df.drop_duplicates(subset= colunas)
print (f'Antes: {antes} , Depois: {df.shape[0]}')
print ('duplicados removidos:' , antes - df.shape[0])

#tratar valores impossíveis

df.loc[df['qtd_produzida'] < 0, 'qtd_´produzida'] = None
df.loc[df['temperatura_c'] >200, 'temperatura_c'] = None
print("valores impossíveis tratados")

df

```
